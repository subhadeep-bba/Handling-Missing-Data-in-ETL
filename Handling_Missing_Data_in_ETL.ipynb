SECTION A – THEORETICAL QUESTIONS

1. ANS - The most common reason for missing data in ETL (Extract, Transform, Load) pipelines is schema drift or uncoordinated changes in upstream source systems, where modified, renamed, or dropped fields cause extraction processes to silently fail or drop incoming rows.

2. ANS - Blindly deleting rows with missing values (often called listwise deletion or dropping NaNs) is considered a bad practice in ETL because it introduces data bias, destroys valuable context, and degrades the performance of downstream machine learning models.

3. ANS -Difference Between Listwise Deletion and Column Deletion
Listwise deletion: Remove rows containing missing values.
Column deletion: Remove columns containing too many missing values.
Appropriate scenario: When working with a small dataset where only a few rows have missing values and removing those rows will not significantly affect the analysis.

4. ANS - Median imputation is preferred over mean imputation for skewed data, such as income, because the median is less affected by extreme values (outliers).

5. ANS - Forward fill (FFill) is a missing-value imputation technique used in data preprocessing and ETL. It replaces a missing value with the most recent non-missing value that appears before it.
Forward fill is most useful for time-series datasets where values are recorded in chronological order and the last known value is a reasonable estimate for the missing period.

6. ANS - Missing values should be flagged before imputation because it helps preserve information about which values were originally missing. Once missing values are replaced, it becomes difficult to identify the original missing records.

7. ANS - Missing income data is not always just a data quality problem. The fact that income is missing can itself provide valuable business insights. It may indicate customer behavior, privacy concerns, or problems in the data collection process.

Why Forward Fill is Suitable Here:-

* **Sequential Continuity:** Forward fill assumes that in a sequential or transactional dataset, the most recent known value is the best estimate for a subsequent missing record until a new value is updated.
* **Preserving Sample Size:** Instead of discarding rows with missing sales data (which would remove almost half the dataset), forward fill maintains full row availability for downstream analytics while providing a logical baseline value.
